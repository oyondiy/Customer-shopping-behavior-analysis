4. Data Cleaning
   ## Column Name Standardization

Column names were cleaned and standardized by:
- Converting all text to lowercase
- Replacing spaces with underscores
- Removing special characters

This ensures compatibility with Python, SQL, and Power BI.

```python
# Step 1: View original columns
print("Original Columns:")
print(df.columns)
```

```python
# Step 2: Clean column names
df.columns = (
    df.columns
    .str.strip()                          # remove leading/trailing spaces
    .str.lower()                          # convert to lowercase
    .str.replace(' ', '_')                # replace spaces with underscore
    .str.replace('(', '', regex=False)    # remove (
    .str.replace(')', '', regex=False)    # remove )
    .str.replace('-', '_')                # replace hyphens with underscore
)
```

```python
# Step 3: View cleaned columns
print("\nCleaned Columns:")
print(df.columns)
```

```python
# step4 Handle duplicates
df.drop_duplicates(inplace=True)
```

```python
# step5 Check for inconsistencies in categorical columns
for col in ['customer_id', 'gender', 'category', 'size', 'season', 'subscription_status', 'shipping_type', 'payment_method', 'frequency_of_purchases']:
    print(col, df[col].unique())
```

```python
# step6 Convert binary categoricals to 0/1
binary_cols = ['subscription_status', 'discount_applied', 'promo_code_used']
for col in binary_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0})
```

## Feature Engineering: Age Group

Customer ages were segmented into categorical groups (Teen, Young Adult, Adult, Middle Age, Senior) to enable better analysis of purchasing behavior across demographics.

```python
# Define age bins and labels
bins = [0, 18, 30, 45, 60, 100]
labels = ['Teen', 'Young Adult', 'Adult', 'Middle Age', 'Senior']

# Create age_group column
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels)

# Check result
df[['age', 'age_group']].head()
```

```python
# . Create ONE powerful feature: average spend per previous purchase
df['Avg_Spend_Per_Prev'] = df['purchase_amount_usd'] / (df['previous_purchases'] + 1)
```

```python
# High rating flag (≥4.5)
df['high_rating'] = (df['review_rating'] >= 4.5).astype(int)
```

```python
# Define mapping from frequency strings to approximate days
frequency_mapping = {
    'Weekly': 7,
    'Fortnightly': 14,
    'Bi-Weekly': 14,
    'Monthly': 30,
    'Quarterly': 91,
    'Every 3 Months': 91,
    'Annually': 365
}

# Create new column
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)

# Check result
print(df[['frequency_of_purchases', 'purchase_frequency_days']].drop_duplicates())
```
